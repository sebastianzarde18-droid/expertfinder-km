## 3. Tabla resumen

```
import json, pandas as pd
filas=[]
for arch in ('parafrasis_resumen.json','parafrasis_resumen_st.json'):
    pass
for arch in ('parafrasis_resumen_llm.json','parafrasis_resumen_st_llm.json'):
    try: d=json.load(open('reports/'+arch))
    except FileNotFoundError: continue
    for k,v in d.items():
        filas.append(dict(configuracion=k, correctas=f"{v['correctas']}/{v['de_con_respuesta']}", falsos_rechazos=v['falsos_rechazos'],
                          rechazos_correctos=f"{v['rechazos_correctos']}/{v['de_sin_respuesta']}", inventadas=v['inventadas'],
                          respuestas_con_llm=v.get('respuestas_con_llm','')))
pd.DataFrame(filas)
```

Revisa que `respuestas_con_llm` sea mayor que 0 en las filas `llm_*`: si es 0, el LLM no se usó (clave o red) y esa fila es en realidad el modo extractivo.
